### Estratégias para Storage e oferta de dados

![Diferenças](https://www.databricks.com/wp-content/uploads/2020/01/data-lakehouse-new.png)

#### Data Warehouse 
Normalmente era construída sobre um banco de dados relacionais, atualmente existem produtos que permitem a construção sobre arquivos. Dentre estes produtos são exemplos o Snowflake e o Redshift. 

Metodologias associadas: Modelagem dimensional (Kimball) para o modelo lógico e Data Vault (Inmon) para o modelo físico. Estas metodologias também podem ser aplicadas a soluções do tipo Lakehouse, porém não é necessária a distinção entre modelos físicos e lógicos.

#### Data Lake
Repositório para dados estruturados, não-estruturados ou semi-estruturados. Pode ser de difícil manutenção e a falta de governança o torna em Data Swamp.

#### Data Lakehouse
Estruturas que replicam boas características dos bancos relacionais e incorporam boas práticas de governança. Dentre os produtos que utiizam desta estrutura são exemplos Databricks e o BigQuery

Possivel Estrutura:
*Ingestão | Armazenamento | Preparo | API | Semântica* <br />

Cada componente pode ser estruturada separadamente ou podemos pensar um fluxo contínuo. As estruturas subjacentes entre cada uma das camadas/componentes podem ser estruturadas de maneiras distintas. Por exemplo, podemos usar Python para a organização do armazenamento/preparo, fazer a ingestão e servir os dados utilizando uma linguagem como C# e construir toda a camada semântica utilizando R.

### Ingestão

#### Slowly Changing Dimension (Operações de Merge)

##### Tipo 01

![Tipo1](https://docs.oracle.com/cd/E41507_01/epm91pbr3/eng/epm/phcw/img/i-4642ab0fn-7fcb.png)

Esta técnica consiste em substituir registros quando há alteração de registros previamente existentes. Tem potencial para ser a mais eficiente mas reduz a capacidade de viagem no tempo e auditoria das mudanças.

##### Tipo 02

![Tipo1](https://docs.oracle.com/cd/E41507_01/epm91pbr3/eng/epm/phcw/img/i-4642ab0fn-7ff3.png)

Esta técnica consiste em duplicar a chave que foi atualizada e a marca toda a tabela com timestamps de validade. Tem potencial para aumentar a governança e permite fazer consultas em versões prévias dos dados

##### Tipo 03

Esta técnica consiste em duplicar a chave que foi atualizada e constrói colunas com o valor prévio do campo. Tem potencial para aumentar a governança e permite fazer consultas em versões prévias dos dados, porém apenas para a alteração mais recente.

### Armazenamento

#### Delta Tables

- ACID compliance: Atomicidade, Consistência, Isolamento e Durabilidade.

Atomicidade: cada instrução de uma transação (leitura, gravação, atualização ou exclusão de dados) é tratada como uma única unidade. Ou as instruções são todas executadas ou nenhuma é executada.

Consistência: garante que as transações apenas modifiquem as tabelas de maneiras predefinidas e previsíveis

Isolamento: vários usuários podem ler e gravar na mesma tabela ao mesmo tempo, mas as transações são isoladas para que as simultâneas não interfiram ou afetem umas às outras.

Durabilidade: garante que as alterações de dados feitas por transações executadas com sucesso sejam preservadas, mesmo em caso de falha do sistema
