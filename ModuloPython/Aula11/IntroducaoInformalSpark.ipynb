### Introdução ao Apache Spark


```python
import os
from pathlib import Path
from datetime import date

from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
os.environ["SPARK_HOME"] = "D:/07.RecursosDesenvolvimento/spark"
os.environ["HADOOP_HOME"] = "D:/07.RecursosDesenvolvimento/spark/hadoop"
```

```python
spark = SparkSession.builder.appName("Introducao ao spark").getOrCreate()
```

```python
spark.conf.set("spark.sql.session.timeZone", "UTC+0")
spark.conf.set("spark.sql.execution.arrow.pyspark.enabled", "true")
```

```python
caminho_base = Path("C:/Users/RenatoFileto/curso_bigdata/ModuloPython/data/raw")
caminho_staging = Path("C:/Users/RenatoFileto/curso_bigdata/ModuloPython/data/staging/base_semana02")
caminho_base_csv = caminho_base.joinpath("base_semana02.csv")
```

```python
base_semana02_1 = spark.read.options(delimiter=";", header=True).csv(caminho_base_csv.__str__())
```

```python
base_semana02 = spark.read.csv(
    caminho_base_csv.__str__(),
    sep=";",
    header=True
)
```

```python
base_semana02_1.show()
```

```python
base_semana02.show()
```

```python
base_semana02 = base_semana02.withColumn(
    "trat_amostra",
    F.col("amostra")
)

base_semana02.select("amostra", "trat_amostra").show()
```

```python
lista_colunas = []

base_semana02.write.partitionBy("amostra", "foto").csv(caminho_staging.__str__(), sep=";", header=True, mode="overwrite")
```

```python
base_semana02_part = spark.read.csv(
    caminho_staging.__str__(),
    sep=";",
    header=True
)
```

```python
base_semana02_part.select("amostra").groupby("amostra").agg(F.count("*")).show()
```
